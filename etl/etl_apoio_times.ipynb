# Imports

```python
import pandas as pd
import numpy as np
```

# Constants

```python
# Raw Data Path
PATH_RAW = "../data/raw/"
# Palpites
FILE_APOIO = "3 - apoio.xlsx"
LIST_SHEET_APOIO = ["paises", "combinações_terceiro_lugar", "mapa_partidas_fg","mapa_partidas_32"]

# Processed Data Path
PATH_PROCESSED = "../data/processed/"
```

# ETL + Save

## Paises

```python
for sheet in LIST_SHEET_APOIO:
    df = pd.read_excel(PATH_RAW + FILE_APOIO, sheet_name=sheet)
    df.to_csv(PATH_PROCESSED + f"apoio_{sheet}.csv", index=False)
```
