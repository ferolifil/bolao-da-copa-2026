# Imports

```python
import pandas as pd
import numpy as np
```

# Constants

```python
# Raw Data Path
PATH_RAW = "../data/raw/"
# Palpites
FILE_GABARITO = "2 - gabarito.xlsx"

# Processed Data Path
PATH_PROCESSED = "../data/processed/"
```

# ETL

```python
# Lendo dados brutos de palpites
df_gab_fg = pd.read_excel(PATH_RAW+FILE_GABARITO, sheet_name="gabarito_fg",engine="openpyxl")
```

```python
df_gab_fg["result_ref_casa"] = np.select(
    [
        (df_gab_fg["vl_time_casa"].isna()) | (df_gab_fg["vl_time_fora"].isna()),  # jogo não ocorreu
        df_gab_fg["vl_time_casa"] > df_gab_fg["vl_time_fora"],   # vitória do mandante
        df_gab_fg["vl_time_casa"] == df_gab_fg["vl_time_fora"],  # empate
    ],
    ["", "V", "E"],  # valores para as duas primeiras condições
    default="D"  # derrota do mandante
)

df_gab_fg
```

# Save

```python
# Salvar sem o índice
file_path = PATH_PROCESSED + "gabaritos_fg_processados.csv"
df_gab_fg.to_csv(file_path, index=False, encoding='utf-8')
```
